# Step 1: Setup

```python
import json

# --- Define two simple agents ---
def planner_agent(task_description):
    """Planner agent converts natural language into a structured task JSON."""
    message = {
        "protocol": "ACP",
        "performative": "request",
        "role": "Planner",
        "intent": "generate_code",
        "task": task_description,
        "language": "python"
    }
    return message

def coder_agent(message):
    """Coder agent receives the JSON task and responds with code + confirmation."""
    if message["intent"] == "generate_code":
        code_snippet = "print('Hello, world!')"  # (simulated output)
        response = {
            "protocol": "ACP",
            "performative": "inform",
            "role": "Coder",
            "status": "completed",
            "task": message["task"],
            "code": code_snippet
        }
        return response
    else:
        return {"error": "Unknown intent"}
```

# Step 2: Simulate a Conversation

```python
# Step 1. Natural language request from human → Planner
user_request = "Write a Python program that prints 'Hello, world!'"
planner_message = planner_agent(user_request)

print("📤 Planner → Coder (ACP JSON):")
print(json.dumps(planner_message, indent=2))

# Step 2. Planner → Coder
coder_response = coder_agent(planner_message)

print("\n📥 Coder → Planner (ACP JSON):")
print(json.dumps(coder_response, indent=2))

# Step 3. Human-readable output
print("\n✅ Human-readable confirmation:")
print(f"Task: {coder_response['task']}\nGenerated Code:\n{coder_response['code']}")
```
