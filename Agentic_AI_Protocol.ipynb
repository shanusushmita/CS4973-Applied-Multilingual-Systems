```python
import json

# Step 1. Natural Language Request
user_request = "Schedule a meeting with Alice and Bob next Tuesday at 3 PM."

# Step 2. Convert request into ACP JSON (simulated LLM output)
acp_request = {
    "protocol": "ACP",
    "performative": "request",
    "intent": "schedule_meeting",
    "participants": ["alice@example.com", "bob@example.com"],
    "time": "2025-09-30T15:00:00Z",
    "location": "Zoom"
}

# Step 3. Simulated agent response (in ACP JSON)
acp_response = {
    "protocol": "ACP",
    "performative": "inform",
    "status": "confirmed",
    "meeting_id": "12345",
    "calendar_link": "https://zoom.us/j/12345"
}

# Step 4. Display everything
print("Natural Language Input:")
print(user_request)

print("\nACP Request Message:")
print(json.dumps(acp_request, indent=2))

print("\nACP Response Message:")
print(json.dumps(acp_response, indent=2))

print("\nHuman-Readable Confirmation:")
print(f"✅ Meeting scheduled with {', '.join(acp_request['participants'])} "
      f"on {acp_request['time']} at {acp_request['location']}. "
      f"Link: {acp_response['calendar_link']}")
```
